## Accuracy will lie / Recision & Recall tradeoff

model can have high Accuracy but perform poorly in Precision and Recall

Accuracy only cares about how many predictions were correct overall — it doesn't care about which predictions were correct, or whether those were the easy cases or the hard cases.
When the negative class dominates the dataset (like 98 healthy people vs 2 sick people), a model can rack up a high accuracy score just by nailing the easy, majority negative cases — it doesn't even need to understand what makes a positive case different. Meanwhile, the minority positive class (the actual sick patients, the ones that matter most) can be completely ignored, and accuracy won't reflect that failure at all.
That's the core problem: accuracy treats every correct prediction as equally valuable, but in reality, correctly catching a rare positive case (the sick patient) is far more important than correctly predicting another obvious negative case (a healthy person). Accuracy is blind to that distinction — which is why you need Recall and Precision to see what's actually happening underneath the surface.

## ROC curve 
ROC Reveriver Operator Characteristic

A probability value lies between 0 and 1. Zero means that there is no probability of occurrence and one means that the occurrence is certain.
But while working with real-time data, it has been observed that we seldom get a perfect 0 or 1 value.
There comes the concept of Threshold. A threshold is set, any probability value below the threshold is a negative outcome, and anything more than the threshold is a favourable or the positive outcome.
