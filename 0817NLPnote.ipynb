2. Bags of Words Model
只统计每个词在文本中出现的次数/频率。
But if you only count the occurance, frequency, the stops words are given too much weight
问题：只看频率的话，停用词（的、是、a、the）出现次数最多，反而被赋予了过高的权重——但它们其实是最没信息量的词。

3.Tf-Idf 
TF（Term Frequency）- occurance times, how many times the word repeately occuring

IDF（Inverse Document Frequency） - 词在所有文档中出现得越普遍，权重就越低
So as to penalise the wrong weight, give more importance to important words.
目的：惩罚那些到处都出现的"废话词"，给真正有区分度、重要的词更高权重。

TF负责"这个词在这篇文档里到底有多常见"，IDF负责"这个词在所有文档里到底有多稀有/独特"，两者相乘，既重视高频词，又惩罚"到处都是"的废话词

above all rely on the occurance frequency of the individual word

注意：One-Hot、BoW、TF-IDF 这三者本质上都只依赖单个词的出现频率，没有真正理解词序和上下文。

## 序列模型的演进（为了处理"顺序"和"上下文"）
6. RNN （Recurrent Neural Network，循环神经网络）
it connects to itself recurrently
让网络的输出循环连接回自己，从而记住之前处理过的信息，捕捉上下文和顺序。

RNN被设计成：每处理完一个词，就把这个"记忆"传给下一步，一起参与下一个词的处理。
X → h → L
    ↑
   (h会传给下一个h)

拆解一下这几个字母：
X：当前输入的词（比如"我"）
h：隐藏层的"记忆状态"（hidden state）——这是RNN的灵魂，它把"到目前为止看过的所有内容"压缩成一个向量
L：这一步的输出（比如预测下一个词，或者做分类）

7. LSTM （Long Short-Term Memory）

RNN的升级版，专门设计用来解决长序列中"记不住"早期信息的问题。缺点：无法并行处理（只能一步一步串行算）。
普通RNN只有一个简单的"记忆传递"，容易导致记忆传着传着就"失真/丢失"（这叫梯度消失问题）。LSTM的解决办法是：在每一步都加装三个"阀门"
input gate（输入门）：决定"这次新信息要不要写入记忆"
forget gate（遗忘门）：决定"要不要把旧记忆的一部分清空"
output gate（输出门）：决定"当前记忆里，哪部分要拿出来用"

类比：普通RNN的记忆像一杯不断被兑水的果汁，喝到后面味道越来越淡（记忆丢失）。LSTM相当于给这杯果汁装了三个可调节的阀门——一个控制"倒多少新果汁进去"，一个控制"倒掉多少旧的"，一个控制"倒出多少给你喝"。这样调配出来的果汁浓度能保持得更稳定、更久（长期记忆更好）。

no parallel processing

8. GRU (Gated Recurrent Unit)
gated recurrent unit
LSTM的简化版，结构更精简，效果接近但计算更快。

GRU（Gated Recurrent Unit）是LSTM的简化版——把三个阀门砍成两个：
    Reset gate（重置门）：决定"要不要忽略掉之前的记忆，主要看当前输入"
    Update gate（更新门）：决定"新记忆和旧记忆各占多大比例混合"（图里那个 1- 说明这两个比例是互补的——更新多少新的，就相应保留多少旧的，加起来正好是100%）

跟LSTM的三门比，GRU砍掉了单独的output gate，把"读取"和"更新"的逻辑合并简化了。

效果：结构更简单 → 参数更少 → 训练更快、更省内存，实际效果通常跟LSTM相差不大，很多场景下可以直接替代LSTM使用。

the above 3 -RNN/LSTM/GRU  have drawbacks
1. long context
2. serial processing (not paralell)
3. high memory
4. too complex models

RNN / LSTM / GRU 共同的痛点：
    长上下文处理能力弱（句子一长，前面的信息容易"忘记"）
    只能串行处理，无法并行，训练速度慢
    显存/内存占用高
    模型结构复杂，难以优化和扩展

## so people come up with  Encoder-Decoder & Transformer

natural language genration evolved

transformer -- encoder + decoder along with one major componenet called- multi head attention - fixing
     long context by parallel processing

为了解决上面这些问题，人们提出了 Encoder-Decoder（编码器-解码器）架构，这也推动了自然语言生成（NLG）的发展。
而真正的突破是 Transformer：
    Transformer = Encoder + Decoder + 一个革命性的核心组件 —— Multi-Head Attention（多头注意力机制）
Multi-Head Attention 解决的关键问题：
    ✅ 支持并行处理（不用像RNN那样一步步来，训练速度大幅提升）
    ✅ 能更好地捕捉长距离上下文关系
这就是为什么 Transformer 成了现代所有大语言模型（GPT、Claude等）的基础架构。

above are basic NLP concepts

之前的RNN/LSTM只能做"输入一个词，输出一个词"这种长度对应的任务。但翻译任务的输入输出长度往往不一样（比如17个中文字，翻译成11个英文词）。Encoder-Decoder架构解决了这个"长度不匹配"的问题——Encoder只管把任意长度的输入"压缩"成一个固定大小的向量，Decoder再从这个向量"解压"出任意长度的输出，输入输出长度完全可以不同。

但这个架构本身也有个明显短板

所有的信息都被压缩进了那"一个"向量（图里那几条曲线汇聚到的那一个点）。如果原句很长（比如一整段话），这一个向量根本装不下那么多细节，会导致信息丢失——句子越长，翻译质量往往越差。

这正是你笔记里下一步要提到的"Multi-Head Attention"要解决的问题：与其把所有信息硬塞进一个向量，不如让Decoder在生成每一个词的时候，都能"回头看"一遍Encoder里的所有中间状态，自己决定"这一步该重点参考原句里的哪几个字"——这就是Attention机制的核心思想，也是Transformer相比传统Encoder-Decoder的革命性突破。
