## Boosting

    not parallel, in series

    Boosting is an ensemble approach(meaning it involves several trees) that starts from a weaker decision and keeps on building the models such that the final prediction is the weighted sum of all the weaker decision makers.

  Ensemble parameters are calculated in stagewise way which means that while calculating the subsequent weight, the learning from the previous tree is
  considered as well.

## Ada (mannually, not used much)

adaptive boosting : first weaker decision tree - manually adjust the weight - pass to the second - manually adjust the weight - third decision tree - .... - final weight

## Gradient boosting

first tree - next model (apply gradient descend) - new weight
you don't manually change the weight, apply gradient descend

but if the data is imbalance, high variance, then the models might fail

** so we have extreme gradient boosting
