***  we will mainly talk about Supervised model/ supervision
        Regression
        Linear regression

Linear Regression is one of the most fundamental algorithms in the Machine Learning world.
Discrete/continuous independent variables
A best-fit regression line
Continuous dependent variable

（features / Independent Variables） predict （label / Dependent Variable）

IV DV: explanatory, stats
machine learning is not to explain, is to predict, so they use the name "feature" & "label"
Feature- pattern recognition - label  (accuracy、loss）

Feature Selection

Residual:    actual y - y^ predicted y 

Y= a +b*x +e
A intercept
B slope
E error term/ residual

Machine learning goal is to let model find the best b slope value，make the linear regression line the best fit line上，minimize the residual.


**** So how to do it?
1.	OLS ordinary least square method  
SSE= summation  (y - y^)squared, total error squared (when sample volume is small, get formula, get the value)
2.	Gradient descent (when the data volume is huge).  
an optimizer, small steps by steps, incremental steps

Find best-fit line, with least residuals
** R² and Adjusted R²
R²  to judge 
R2 = 0.75; our model fits 75% of the total data set
R²=1 , then  it represents 100%

but there is a bug with R², as the IVs increase, R² will definetely increase, even when new IVs are irrelavant to DV

So we need Adjusted R²
Adjusted R2 (when there are increased independent variable, the R2 increases as well, but it doesn’t mean that the new IVs have any correlation with the DV)
Adjusted R2 can reduce the weight of the unrelated IVs
