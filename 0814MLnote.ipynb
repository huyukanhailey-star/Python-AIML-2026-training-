## K means clustering

K-Means is a clustering approach in which the data is grouped into K distinct non-overlapping clusters based on their distances from the K centres.

k number of random points

you initialize the k (how many clusters),
the model will find k centoid (central points) in random position
it will calculate the distance of the points to centroid, assign the near points to centroid
recaluculate and reset central points
repeat until you find the final central points

the clusters will have similar patterns 

non-overlapping, one point belongs to one cluster

but a drawback of K-means, you need to set a k number,
so we have Elbow Method
to choose the best K value for the clusters

WCSS(within-cluster sum of squared distances) - graph

but sometimes WCSS won't work, no obvious point on the graph

so we need Silhouette score to set K

## Silhouette score

how well the cluster is formed?

how similar a data point is to its own cluster vs other cluster
separate from others, close to the similar ones in its own cluster

Two key distances

For every data point, we calculate two things:​

 (intra-cluster distance): the average distance from this point to all the other points in the same cluster
    → smaller is better — it means the point is closely aligned with its own cluster, i.e. it fits in well with "its own people"

(nearest-cluster distance): the average distance from this point to all the points in the nearest neighboring cluster (the next-best alternative cluster it could belong to)
    → larger is better — it means the point is well-separated from other clusters, i.e. it's far from "outsiders"

"distance from outsiders" minus "distance from own group."
The bigger this gap, the more confidently this point belongs to its assigned cluster.

-1 to +1
we want +1
0 （point is near cluster boundaries or clusters overlap ，then check why)
-1 poor cluster
