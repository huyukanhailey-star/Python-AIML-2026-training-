## How to judge "pure" or not?

1. Entropy 
calculate the randomness/"impurity" in the node data
using : log
if the same, Entropy = 0, pure
if half half, Entropy = 1, most random,impure

2. Information Gain
calculate the decrease in the entropy after spliiting a node
before after

* the above two take too much calculation, log takes a lot
* so we have another one

3. Ginni Impurity 
just like Entropy, but faster and simpler
'Gini impurity is a measure of how often a randomly chosen element from the set would be incorrectly labelled if it was randomly labelled according to the distribution of labels in the subset! 

Gini=0, pure
Gini = 0.5 50/50

## Greedy Approach

From the root node, go over every feature, every possible splitting point, 
calculate the Gini after splitting, choose the purer one
until the leaf node is pure

the tree will grow too much 
"The process has a high chance of overfitting"
and then model will be overfit

to solve this problem, we need Pruning

Advantages of Decision Tree:
• It can be used for both Regression and Classification problems.
• Decision Trees are very easy to grasp as the rules of splitting is clearly mentioned.
• Complex decision tree models are very simple when visualized. It can be understood just by visualising.
• Scaling and normalization are not needed.

Disadvantages of Decision Tree:
• A small change in data can cause instability in the model because of the greedy approach.
• Probability of overfitting is very high for Decision Trees.
• It takes more time to train a decision tree model than other classification

## Cross Validation
how you split your data for testing & training
how much used for training, how much used for testing

1. Hold-out
80% for training, 20% for testing
computationally less costly
but "depends heavily on which data points end up in the training set and which in test data"

2. K-Fold
divide the whole data set into k sets of equal sizes
can handle variance issue
but costly computation

3. LOOCV
Leave One out cross validation
repeat n times, still costly
"LOOCV is a special case of k-fold CV, where k becomes equal to n"

4. stratify split
maintain same ratio when deleaing with train data & test data
to handle overfitting issue/ imbalance data

mostly using hold-out method

Cross Validation: can be applied to any type of models, linear regression, logitic regression, decision tree
