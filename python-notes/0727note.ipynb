```python
# __init__ method 


class Person:
    def __init__(self, name, surname, year_of_birth):    # self refers to the instance of the class
        self.name = name
        self.surname = surname
        self.year_of_birth = year_of_birth

alec_obj = Person("alec", "Baldwin", 1958)

print(alec_obj)
print("%s %s was born in %d." % (alec_obj.name, alec_obj.surname, alec_obj.year_of_birth))
```

```python
class Person:
    def __init__(a, name, surname, year_of_birth):
        a.name = name
        a.surname = surname
        a.year_of_birth = year_of_birth

    def age(a, current_year):
        return current_year - a.year_of_birth  # calculates the age based on the current year

    def summ(a, current_year):
        return current_year + a.year_of_birth

    def __str__(a):
        return "%s %s was born in %d ." % (a.name, a.surname, a.year_of_birth)   # __str__

alec = Person("Alec", "Baldwin", 1997)
print(alec)
print(alec.age(2026))     
print(alec.summ(2026))
```

```python
class Person:
    def __init__(self, name, surname, year_of_birth):
        self.__name = name              # _ _ double underscore for private attribute
        self.__surname = surname
        self.__year_of_birth = year_of_birth

alec = Person("Alec", "Baldwin", 1958)

print(alec.__name)
```

```python
class Person:
    def __init__(self, name, surname, year_of_birth):
        self.__name = name              # _ _ double underscore for private attribute
        self.__surname = surname
        self.__year_of_birth = year_of_birth

alec = Person("Alec", "Baldwin", 1958)
print(alec._Person__name) 
```

```python
class Student(Person):
        def __init__(self, student_id, *args):
            super(Student, self).__init__(*args)
            self._student_id = student_id

charlie = Student(1, 'Charlie', 'Brown', 2006)
print(charlie._student_id)
print(charlie._Person__name)
print(charlie._Person__surname)
print(charlie._Person__year_of_birth)
```
