Encapsulation composition
one class containing another class

```python
class Battery:
    def charge(self):
        print("charging...")

class Phone:
    def __init__(self):
        self.battery = Battery()   # wrap the Battery instance inside the Phone

    def turn_on(self):
        print("turning on...")
        self.battery.charge()    
        print("turned on！")

my_phone = Phone() 
my_phone.turn_on() 
```

```python
class Text:
    def render(self):
        return "Hello"

x = Text()
x.render()   

class BoldWrapper:
    def __init__(self, wrapped):
        self.wrapped = wrapped        # save the passed-in object

    def render(self):
        return f"<b>{self.wrapped.render()}</b>"

simple = Text()              # simple.render() → "Hello"

bold = BoldWrapper(simple)   # simple into BoldWrapper 
bold.render()
```

Polymorphism & Duck Typing

```python
a = 5
a = 'ght'
a = 6.8
a = [1, 2, 3]
print(type(a))    # Duck Typing, no matter what type `a` is, we can assign different types to it
```

```python
def summer(a, b):
    return a + b                # Polymorphism, the same function works for different types

print(summer(1, int("1")))              # 2         ← two int adding
print(summer(["a","b","c"], ["d","e"]))  # ['a','b','c','d','e']  ← two lists adding
print(summer("abra", "cadabra"))         # abracadabra  ← two strings adding
```

Instance method:
static method
@classmethod cls

```python
class Library:
    total_books = 0          # class variable

    def __init__(self, title):
        self.title = title    # instance variable
        Library.total_books += 1   

    # 1. Instance method  # used to operate on instance-level data
    def describe(self):
        print(f"title:{self.title}")

    # 2. classmethod  # used to operate on class-level data
    @classmethod
    def show_total(cls):
        print(f"there are {cls.total_books} new books")

    # 3. staticmethod ——  a method that does not operate on instance-level or class-level data
    @staticmethod
    def is_valid_isbn(isbn):
        return len(isbn) == 13

b1 = Library("Lord of Ring")
b2 = Library("Harry Potter")

b1.describe() # instance method
b2.describe()   

Library.show_total()  # class method

print(Library.is_valid_isbn("9783161484100"))  # static method check ISBN
```

```python
class Math:
    @staticmethod
    def add(a, b):
        return a + b

print(Math.add(5, 7))   # 12
```
