```python
## Simple Linear regression is a method for predicting a quantitative response 
using a single feature ("input variable").

feature_cols = ['TV']
X = data[feature_cols]
y = data.sales

from sklearn.linear_model import LinearRegression
lm = LinearRegression(fit_intercept=False)
lm.fit(X, y)

print(lm.intercept_)
print(lm.coef_)
```

```python
## Multiple Linear Regression

Till now, we have created the model based on only one feature. 
Now, we'll include multiple features and create a model to 
see the relationship between those features and the label column. 
This is called Multiple Linear Regression

feature_cols = ['TV', 'radio', 'newspaper']
X = data[feature_cols]
y = data.sales

lm = LinearRegression()
lm.fit(X, y)

print(lm.intercept_)
print(lm.coef_)
```
